```python
import time
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')

frases = ["O universo é finito?", "Como funcionam redes neurais?"] * 500  # simula carga

start = time.time()
embeddings = model.encode(frases, batch_size=32, show_progress_bar=True)
print(f"Tempo: {time.time() - start:.2f}s  Shape: {embeddings.shape}")
```
