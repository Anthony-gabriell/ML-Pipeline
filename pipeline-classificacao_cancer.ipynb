```python
from sklearn.datasets import load_breast_cancer # carrega dataset de cancer
from sklearn.model_selection import train_test_split # divide os dados de treino e teste
from sklearn.preprocessing import StandardScaler #normaliza as features, coloca tudo na mesma escala
from sklearn.linear_model import LogisticRegression # primeiro modelo que vamos treinar
from sklearn.tree import DecisionTreeClassifier # segundo modelo arvore de decisao
from sklearn.ensemble import RandomForestClassifier # terceiro modelo floresta aleatoria
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
```

```python
# carregando o dataset
cancer = load_breast_cancer()

# analise x = feature e y = label
x = cancer.data # .data acessa a matriz de features do objeto
y = cancer.target # .acessa os rotulos a repostata correta

# exploracao basica
print("Shape de x:", x.shape)
print("Shape de y:", y.shape)
print("Classes:", cancer.target_names)
print("Exemplo de y", y[:10])
```

```python
# split de treino

x_train, x_test, y_train, y_test = train_test_split(
    x,y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Treino:", x_train.shape)
print("Teste:", x_test.shape)
```

```python
# normalizacao
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train) # aprende a escala dos dados de terino, calcula a media e o desvio
x_test = scaler.transform(x_test) # aplica a escala que foi aprendida no trieno ao teste, nada novo

print("Media após escala:", x_train.mean().round(4))
print("Desvio após escala:", x_train.std().round(4))

#normalizacao funciona quando o rsultado fica entre 0 e 1
```

```python
# treino dos modelos primeiro regressao logisitca
# Regressão logística resolve um problema matemático iterativamente. 1000 iterações garante que ele converge
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(x_train, y_train)

# modelo 2 - arvore de decisao
# Limita a profundidade da árvore de decisão. Sem limite ela cresce até decorar todos os dados
dt = DecisionTreeClassifier(max_depth=10, random_state=42)
dt.fit(x_train, y_train)

# modelo 3 - random forest
# Quantas árvores o Random Forest vai construir e combinar. 100 é o padrão
rf = RandomForestClassifier(n_estimators=100)
rf.fit(x_train, y_train)

print("3 modelos treinados")
```

```python
# avalicao do modelo predicao dos 3 modelos
# Cada modelo olha para as 114 amostras de teste que nunca viu durante o treino e tenta prever o tumor.
y_pred_lr = lr.predict(x_test)
y_pred_dt = dt.predict(x_test)
y_pred_rf = rf.predict(x_test)

# relatorio completo de cada modelo
# classification_report gera a tabela completa com precision, recall e F1 para cada classe
print("=== REGRESSÃO LOGÍSTICA ===")
print(classification_report(y_test, y_pred_lr))

print("=== ÁRVORE DE DECISÃO ===")
print(classification_report(y_test, y_pred_dt))

print("=== RANDOM FOREST ===")
print(classification_report(y_test, y_pred_rf))
```
